```python
from pathlib import Path
import glob, re
```

```python
fls = glob.glob('data/*')
```

```python
[re.sub(r'\.csv', '_2009-21.csv', f) for f in fls]
```

```python
for f in fls:
    new_f = re.sub(r'\.csv', '_2009-21.csv', f)
    old_name = Path(f)
    new_name = Path(new_f)

    old_name.rename(new_name)
    print(f"Renamed '{old_name}' to '{new_name}'")
```
